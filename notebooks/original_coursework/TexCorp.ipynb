TexCorp LLC produces two competing products, X and Y. The company would like to sell these products to two separate groups of customers. Customers place different values on a unit of X versus Y (please see the table). Each customer will buy either product X or product Y, but not both. A customer is willing to buy product X if she believes that the premium of product X (its value minus its price) is greater than or equal to the premium of product Y (its value minus its price) and premium of product X is greater than or equal to 0. Likewise, a customer is willing to buy Y if she believes the premium of product Y is greater than or equal to the premium of product X and the premium of product Y is greater than or equal to 0. There are 2000 potential customer in group 1, and 2500 in group 2. TexCorp wants to set prices for each product to ensure that customers from group 1 purchase product X and customers from group 2 purchase product Y. How TexCorp can maximize its revenue?

Algebraic Representation:

Decision Variables:

Px = price of X
Py = price of Y

Given:
N1 = 2000 in Group 1
N2 = 2500 in Group 2

V1x = 10
V1y = 14
V2x = 12
V2y = 17

Objective:
Max = 2000 * Px + 2500 * Py

Contraints:
V1x - Px ≥ V1y - Py
V2y - Py ≥ V2x - Px
V2y - Py ≥ 0

```python
import gurobipy as gp
from gurobipy import GRB
```

```python
# Create Model

model = gp.Model("TexCorp")
```

```python
# Decision Variables

Px = model.addVar(lb=0, name="Px")
Py = model.addVar(lb=0, name="Py")
```

```python
# Objective Function

model.setObjective(
    2000 * Px + 2500 * Py,
    GRB.MAXIMIZE)
```

```python
# Constraints

model.addConstr(10 - Px >= 14 - Py, name="G1_pref_X")
model.addConstr(10 - Px >= 0, name="G1_nonneg")

model.addConstr(17 - Py >= 12 - Px, name="G2_pref_Y")
model.addConstr(17 - Py >= 0, name="G2_nonneg")
```

```python
# Optimize Model
model.optimize()

# Print Results
print(f"Optimal price for X: {Px.X}")
print(f"Optimal price for Y: {Py.X}")
print(f"Maximum revenue: {model.ObjVal}")
```
