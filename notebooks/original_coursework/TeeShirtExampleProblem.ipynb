You are in charge of organizing weekly social events
at your university’s Student Union. These events are a
big draw for students and help build campus
community — especially when free t-shirts are given
out. Your job is to plan how many events to hold each
week during a 6-week period.
In the past, you’ve found that giving out free t-shirts at
events greatly increases student attendance — but
students start to show up in smaller numbers if too
many events are scheduled in a single week.
Specifically, attendance starts to drop by 5% per
event after the second event in a week. For the first two events, you can expect 200
students per event. After that, attendance drops by 5% for each additional event. So for
the third event, you expect 190 students, for the fourth event, 180 students, and so on.
You’ve decided to hold between 3 and 10 events per week, depending on how many t-
shirts you can afford to distribute. Each student receives one free t-shirt at the event, and
you plan to order one shirt for each expected attendee.
<br>
The Student Union has a partnership with three local vendors who supply printed t-shirts:
<br>
• Campus Threads charges $7.00 per shirt, and can provide up to 1,200 shirts per
week.
<br>
• Shirt Station charges $6.50 per shirt, and can provide up to 800 shirts per week.
<br>
• Tee Express charges $8.25 per shirt, but has no limit on how many shirts it can
supply.

```python
base_attendance = 200
drop_rate = .95
```

```python
# Vendors
vendors = [
    {"name": "Shirt Station", "cost": 6.50,"capacity":800},
    {"name": "Campus Threads", "cost": 7.00,"capacity":1200},
    {"name": "Tee Express", "cost": 8.25,"capacity":float("inf")}
]
```

```python
# Calculate total t-shirts needed with declining attendance
def TotalTshirts(events_per_week):
    total_shirts = 0
    for i in range(1, events_per_week + 1):
        if i <= 2:
            attendees = base_attendance
        else:
            attendees = base_attendance * (drop_rate ** (1-2))
        total_shirts += attendees
    return round(total_shirts)
```

```python
def EstimateTeeCost(TeesNeeded):
    remaining = TeesNeeded
    total_cost = 0
    allocation = {}

    for vendor in vendors:
        supply = min(remaining, vendor["capacity"])
        cost = supply * vendor["cost"]
        allocation[vendor["name"]] = str(round(supply)) + '@$' + str(vendor["cost"])
        total_cost += cost
        remaining -= supply
        if remaining <= 0:
            break
    return {
        "Tshirts Needed": TeesNeeded,
        "total_cost": round(total_cost, 2),
        "allocation": allocation
    }
```

```python
TotalTshirts(5)
```

```python
EstimateTeeCost(2500)
```

```python
EstimateTeeCost(TotalTshirts(20))
```
