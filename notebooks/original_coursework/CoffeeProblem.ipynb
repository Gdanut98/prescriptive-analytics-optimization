```python
import gurobipy as gp
from gurobipy import GRB
```

```python
# Create Model

model = gp.Model("BaytownCoffee")

roasts = ["Light", "Medium", "Dark"]
countries = ["Brazil", "Guatemala", "Colombia"]
```

```python
# Profit per pound for each roast

profit = {
    "Light": 2.30,
    "Medium": 1.15,
    "Dark": 2.70
}

# Inverntory per country

inventory = {
    "Brazil": 9000,
    "Guatemala": 6000,
    "Colombia": 7000
}

# Proportion of each bean in each roast

proportions ={
    "Light": {"Brazil": .21, "Guatemala": .39, "Colombia": .40},
    "Medium": {"Brazil": .50, "Guatemala": .29, "Colombia": .21},
    "Dark": {"Brazil": .40, "Guatemala": .21, "Colombia": .39}
}
```

```python
# Decision variables
x = model.addVars(roasts, vtype=GRB.CONTINUOUS,lb = 0, name = "Roast")
```

```python
# Objective: Maximize total profit
model.setObjective(gp.quicksum(profit[r] * x[r] for r in roasts), GRB.MAXIMIZE)
```

```python
# Constriants: Inventory for each country
contraints = {}
for c in countries:
    contraints[c] = model.addConstr(
        gp.quicksum(proportions[r][c] * x[r] for r in roasts) <= inventory[c],
        name = f"Inventory_{c}"
    )
```

```python
# Optimize the model
model.optimize()

# Output
if model.status == GRB.OPTIMAL:
    print("\nOptimal Production Plan:")
    for r in roasts:
        print(f"{r}: {x[r].X:.2f} lbs to Profit: ${profit[r] * x[r].X:.2f}")
    print(f"\nTotal Profit: ${model.ObjVal:.2f}")
else:
    print("No Optimal solution.")
```
